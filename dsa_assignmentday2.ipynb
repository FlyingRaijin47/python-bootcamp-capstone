```python

sales = [45000, 82000, 120000, 67000, 95000]
target = 75000

# 1. SALES PERFORMANCE ANALYZER
print("SALES PERFORMANCE ANALYZER")
print("=" * 50)

def calculate_total(sales):
    total_sales = sum(sales)
    return total_sales

total_sales = calculate_total(sales)
average_sales = total_sales / len(sales)

print("Total Sales:", total_sales)
print("Average Sales:", average_sales)


# 2. HIGHEST AND LOWEST SALES
print("\nHIGHEST AND LOWEST SALES")
print("=" * 50)

highest_sales = max(sales)
lowest_sales = min(sales)

print("Highest Sales:", highest_sales)
print("Lowest Sales:", lowest_sales)


# 3. SALES CLASSIFICATION
print("\nSALES CLASSIFICATION")
print("=" * 50)

for sale in sales:

    if sale >= 100000:
        classification = "Excellent"

    elif sale >= target:
        classification = "Reached Target"

    else:
        classification = "Below Target"

    print(f"Sales: {sale:,} | Classification: {classification}")


# 4. TARGET ACHIEVEMENT
print("\nTARGET ACHIEVEMENT")
print("=" * 50)

target_count = 0

for sale in sales:

    if sale >= target:
        target_count += 1

print("Target:", target)
print("Number of Sales Reaching Target:", target_count)
```
