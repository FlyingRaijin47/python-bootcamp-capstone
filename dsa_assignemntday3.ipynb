```python
import pandas as pd
from pathlib import Path

# Define raw data directory
DATA_DIR = Path("data/raw")

# Load datasets
customers = pd.read_csv(DATA_DIR / "customers.csv")
products = pd.read_csv(DATA_DIR / "products.csv")
orders = pd.read_csv(DATA_DIR / "orders.csv")
order_items = pd.read_csv(DATA_DIR / "order_items.csv")
payments = pd.read_csv(DATA_DIR / "payments.csv")
shipments = pd.read_csv(DATA_DIR / "shipments.csv")
returns = pd.read_csv(DATA_DIR / "returns.csv")

# Organize datasets
datasets = {
    "customers": customers,
    "products": products,
    "orders": orders,
    "order_items": order_items,
    "payments": payments,
    "shipments": shipments,
    "returns": returns
}

# Inspect dataset dimensions
for name, df in datasets.items():
    print(f"{name}: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

```python
# Payment dataset overview
print("Shape:", payments.shape)
display(payments.head())

print("\nData types:")
print(payments.dtypes)

print("\nMissing values:")
print(payments.isnull().sum())

print("\nPayment methods:")
print(payments["payment_method"].value_counts(dropna=False))
```

```python
# Create a copy for Day 3 analysis
payment_analysis = payments.copy()

# Convert amount_paid to numeric
payment_analysis["amount_paid"] = pd.to_numeric(
    payment_analysis["amount_paid"],
    errors="coerce"
)

# Verify conversion
print("Data type:", payment_analysis["amount_paid"].dtype)
print("Missing amounts:", payment_analysis["amount_paid"].isnull().sum())

display(payment_analysis.head())
```

```python
print("PAYMENT FILTERING")
print("=" * 50)

payment_filtered = payment_analysis[
    (payment_analysis["payment_method"].isin(["GCash", "PayMaya"])) &
    (payment_analysis["amount_paid"] > 1500)
].copy()

print("Filtered payment records:", len(payment_filtered))
print("-" * 50)

print("SORTED PAYMENT RECORDS")
print("=" * 50)

payment_filtered = payment_filtered.sort_values(
    "amount_paid",
    ascending=False
)

print(payment_filtered)
```
