```python
# Business Question 4: Monthly Revenue

df = fact_sales.copy()
df['line_revenue'] = df['quantity']*df['unit_price_at_purchase']

df['month'] = (
    df['order_datetime'].dt.to_period('M')
                )
df1 = df.groupby('month')['line_revenue'].sum().reset_index()

print(df1)

bq4_result = df1.copy()
```

```python
# Business Question 5: Top 10 Products by Revenue
df = fact_sales.copy()
df['Total_Revenue'] = df['quantity']*df['unit_price_at_purchase']
df = df[df['product_id'] != 'PROD00000']

                
df1 = (df.groupby('product_id')['Total_Revenue'].sum().reset_index()
       .sort_values(by=['Total_Revenue'], ascending=False)
)
print(df1.head(10).to_string(index=False))


bq5_result = df1.head(10).copy()
```

```python
# Business Question 6: Revenue by Category
df = fact_sales.copy()
df['Total_Revenue']= (df['quantity']*df['unit_price_at_purchase'])

df1 = (df.groupby('category')['Total_Revenue']
       .sum()
       .reset_index()
       .round(2)
       .sort_values(by='Total_Revenue', ascending= False)
)

print(df1.to_string(index=False))

bq6_result = df1.copy()
```

```python
# Business Question 7: Revenue by Region and City

df = fact_sales.copy()

df['Total_Revenue'] = (
    df['quantity'] * df['unit_price_at_purchase']
)

df1 = (
    df.groupby(['region_name', 'city'])['Total_Revenue']
      .sum()
      .reset_index()
      .round(2)
      .sort_values(by='Total_Revenue', ascending=False)
)

print(df1)

bq7_result = df1.copy()
```

```python
# Business Question 8: Repeat vs New Customers

df = fact_sales.copy()
df = df[df['customer_id'] != 'CUST999999']

df1 = (
    df.groupby('customer_id')['order_id']
      .nunique()
      .reset_index(name='order_count')
)

df1['customer_type'] = (
    df1['order_count']
    .apply(lambda order_count: 'Repeat' if order_count>= 2 else 'New')
)

df2 = (
    df1.groupby('customer_type')['customer_id']
       .count()
       .reset_index(name='customer_count')
)

print(df2)

bq8_result = df2.copy()
```
